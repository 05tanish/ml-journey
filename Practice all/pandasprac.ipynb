```python
#basic question
#1.Array ka datatype 
print(type(arr))
#2.Array ki dimensions
print(arr.ndim)
#3.Array ki shape
print(arr.shape)#--->rows*coloumn
#4.Array me total elements
print(arr.size)
#5.Memory size (bytes)
print(arr.nbytes)
```

```python
#6.First row print
print(arr[0])
#7.Last row print
print(arr[-1])
#8.Second column print
print(arr[:,1])
#9.Last column
print(arr[:,-1])
#10.Pehla element
print(arr[0,0])
```

```python
#11.Element 33 access
print(arr[2,2])
#12.First 2 rows
print(arr[:2])
#13.First 3 columns
print(arr[:,:3])
#14.Bottom-right 2×2 matrix
print(arr[2:,2:])
#15.Array ko transpose
arr.T
```

```python
#16.Array ka Sum
print(arr.sum())
print(np.sum(arr))
#17.Mean nikalo.
print(np.mean(arr))
print(arr.mean())
#18.Maximum
print(arr.max())
print(arr.min())
#19.Maximum kis Index par
print(np.argmax(arr))
print(arr.argmax())
#20.min
print(arr.argmin())
#21.Har Row ka Sum
print(arr.sum(axis=1))
#22. Har Column ka Sum
print(arr.sum(axis=0))
#23.Har Row ka Maximum
print(arr.max(axis=1))
#24. Sabhi Elements ko 10 se Multiply karo.
print(arr*10)
#25.Square nikalo.
print(np.square(arr))
#26.sqrroot
print(np.sqrt(arr))
```

```python
#27. Logarithm nikalo.
print(np.log(arr))
```

```python
#Indexing & Slicing
#28.Even Rows Print 
print(arr[::2])

#29.odd rows
print(arr[1::2])
#30 even cpoloumn 
print(arr[:,::2])
#31 odd coloumn
print(arr[:,1::2])
#32.diagonal
print(arr.diagonal())
#33.Center ke 4 Elements Print Karo
print(arr[1:3,1:3])
#34.Reverse Rows
print(arr[::-1])
#35.Last Column ko 100 Bana Do
newarr = np.copy(arr)
newarr[:,-1] = 100
print(newarr)
```

```python
#36.Sabhi Odd Numbers ko -1 Bana Do
newarr2 = np.copy(arr)
newarr2[newarr2%2!=0] = -1
print(newarr2)
#37. 30 se Bade Elements Print Karo
print(arr[arr>30])
#38. 20–40 ke Beech ke Elements Print Karo
print(arr[(arr>20) & (arr<40)])
#39.Boolean Mask Banao
mask = arr>30
print(mask)
#40.
```
