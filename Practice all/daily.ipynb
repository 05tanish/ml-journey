```python
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Sales':[100,200,150,250,120,100,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
df
```

```python
cat_sales_sum=df.groupby('Category')['Sales'].sum()
cat_sales_sum
```

```python
sto_sales_sum=df.groupby('Store')['Sales'].sum()
sto_sales_sum
```

```python
multi_sales_sum=df.groupby(['Category','Store'])['Sales'].sum()
multi_sales_sum
```

```python
# convert the multiindex series to a clean dataframe
multi_sales_df=multi_sales_sum.reset_index()
multi_sales_df
```

```python
# calculate multiple descriptive statistics for the sales column
df['Sales'].agg(['mean','median','min','max','std','count','var'])
```

```python
grouped_agg=df.groupby('Category').agg({
    'Sales':['sum','max'],
    'Quantity':['mean','median']
})
grouped_agg
```

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['John','Amaa','Peter','Linda','Bob'],
    'department':['HR','IT','Finance','IT','HR']
})
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[20000,45000,38000,28000,30000],
    'bonus':[5000,2000,3000,3500,2500]
})
print("---Employee Data---\n",employees)
print("---salary data---\n",salaries)
```

```python
pd.merge(employees,salaries,on='employee_id',how='inner') #inner merge: gives the commom data values from both the dataframes
```

```python
pd.merge(employees,salaries,on='employee_id',how='outer') #outer merge :gives all the rows from both the data frames
```

```python
pd.merge(employees,salaries,on='employee_id',how='left')
#left_merge : include all rows from the left dataframes and only matching rows from right dataframes
```

```python
pd.merge(employees,salaries,on='employee_id',how='right')
#right_merge : include all rows from the right dataframes and only matching rows from left dataframes
```
