```python
docs = [
"RAG combines retrieval with generation.",
"LangGraph builds agent state machines.",
"FAISS is a vector similarity search library.",
"Refunds are processed within 5-7 business days.",
"Our return policy allows exchanges within 30 days."
]
```

```python
# import faiss, sentence transformer
import faiss
from sentence_transformers import SentenceTransformer
```

```python
# use allminimlm embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")
doc_embeddings = model.encode(docs, normalize_embeddings = True)
```

```python
# check no. of docs, embeddings shape
print("no of docs = ",len(docs))
print("embeding shape = ",doc_embeddings.shape)
# print first few embeddings for 1 doc
```

```python
# create a query
query = "How do I get my refund "
```

```python
# check the cosine similarty from query to database(doc_embeddings)
query_embeddings = model.encode(query, normalize_embeddings=True)
```

```python
import numpy as np
# Define cosine similarty score
def cosine_similarity(a, b):
  return np.dot(a,b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
# compute the scores for rach doc andd display
for i, emb in enumerate(doc_embeddings):
  score = cosine_similarity(query_embeddings, emb)
  print(docs[i], " : ", score)
```

```python
# we have done embedding for 5 small docs, what if 100's, millions docs
# FAISS to handle this process fast and accurately
# convert the all docs embeddings to 1D array
dim = doc_embeddings.shape[1]
index = faiss.IndexFlatIP(dim)
index.add(doc_embeddings)
print("Total index : ",index.ntotal)
print("Vectro dimension : ",dim)
```

```python
# apply it to query for search
def search(q, k=3):
  q_vec = model.encode([q], normalize_embeddings=True)
  scores, indices = index.search(np.array(q_vec), k)
  print("query = ",q)
  for score, idx in zip(scores[0], indices[0]):
    print(score, docs[idx])
```

```python
query = "Hoe do I perfom similarty search"
search(query)
```
