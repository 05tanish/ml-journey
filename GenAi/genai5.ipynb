```python
from langchain_core.documents import Document

document = Document(
    page_content=raw_text,
    metadata={
        "source": "/content/drive/MyDrive/software_developer (1).pdf"
    }
)

print(document)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=250
)
```

```python

chunks = text_splitter.split_documents([document])

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk.page_content)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS vector database created successfully!")
```

**retrive**

```python
retriver = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
query = "What are my technical skills?"

docs = retriver.invoke(query)

for i, doc in enumerate(docs):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```

Build RAG chain LLM|RAG
chat prompt

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are an AI assistant answering questions about Tanish jain's resume.

Use ONLY the information provided in the context to answer the question.

If the answer is not available in the context, say:
"I don't have enough information in the resume to answer this question."

Context:
{context}
"""
    ),

    MessagesPlaceholder(variable_name="chat_history"),

    (
        "human",
        "{question}"
    )
])
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
model = AutoModelForCausalLM.from_pretrained("distilgpt2")

pipeline_gen = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=512,
    device_map="auto",
)

LLM = HuggingFacePipeline(pipeline=pipeline_gen)
```

```python
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from operator import itemgetter

def get_retrieved_context(input_dict):
    """
    Extracts the question from the input dictionary, retrieves documents,
    and combines their content.
    """
    question_query = input_dict["question"]
    # Ensure retriver gets a string
    retrieved_docs = retriver.invoke(question_query)
    return combine_chunks(retrieved_docs)

chain = (
    {
        "context": RunnableLambda(get_retrieved_context),
        "question": itemgetter("question"),
        "chat_history": itemgetter("chat_history")
    }
    | prompt
    | LLM
    | StrOutputParser()
)
```

Add Memory

```python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
store = {}

def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]
```

```python
# existing chain ko memory ke saath wrap karo
```

```python
chain_with_memory = RunnableWithMessageHistory(chain,
    get_session_history,
    input_messages_key="question",
    history_messages_key="chat_history"
)
```

```python
def combine_chunks(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
response = chain_with_memory.invoke(
    {
        "question": "What projects have I worked on?"
    },
    config={
        "configurable": {
            "session_id": "user_1"
        }
    }
)

print(response)
```
