### Tokenization with Qwen 0.5B

```python
print(f"Using device: {device}")

qwen_0_5_tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen1.5-0.5B")
qwen_0_5_model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen1.5-0.5B").to(device)

print("Qwen 0.5B model and tokenizer loaded.")
```

```python
sentences = [
    "the unbelievable transformer architecture revolutionized the AI, NLP, GenAI",
    "I can't tell tokenization can handle subwords better"
]

for sentence in sentences:
  qwen_tokens = qwen_0_5_tokenizer.tokenize(sentence)
  print('Qwen 0.5B token:', qwen_tokens)
  print()
```

### Text Generation with Qwen 0.5B (Temperature Control)

```python
set_seed(42)
prompt = "The ocean at sunset is"
inputs = qwen_0_5_tokenizer(prompt, return_tensors="pt").to(device)

out = qwen_0_5_model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.1
)
print(qwen_0_5_tokenizer.decode(out[0], skip_special_tokens=True))
```
