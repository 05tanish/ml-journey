```python
from transformers import pipeline
import torch

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    torch_dtype=torch.bfloat16,
    device_map="auto"
)
```

```python
# Direct LLM call
prompt = "Explain Artificial Intelligence in simple words."

result = generator(
    prompt,
    max_new_tokens=100
)

print(result[0]["generated_text"])
```

3 chat prompt templete
write prompt, convert it to langchain chatprompt templet,
before invoking the responce, chain the prompt with llm
invoke the responce

```python
# from langchain_huggingface import HuggingFacePipeline
from langchain_huggingface import HuggingFacePipeline
from transformers import pipeline
from langchain_core.prompts import ChatPromptTemplate
LLM = HuggingFacePipeline(pipeline=generator)
# prompt = "Explain Artificial Intelligence in simple words."
chat_prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI assistant."),
    ("human","Explain{topic}in simple words."),
])
# chain the prompt with llm
chain = chat_prompt | LLM
# invoke the responce
result = chain.invoke({"topic":"Artificial Intelligence"})
print(result)
```

```python
# want to keep only the responce
if "Assistant:" in result:
    result = result.split("Assistant:")[1]
print(result)
```

# adding output parser
- StrOutputParser
- it extract string output form the output of llm
-it gives clean string that is easier to use in next step of the chain



```python
# from langchain_huggingface import HuggingFacePipeline
from langchain_huggingface import HuggingFacePipeline
from transformers import pipeline
from langchain_core.prompts import ChatPromptTemplate
LLM = HuggingFacePipeline(pipeline=generator)
# prompt = "Explain Artificial Intelligence in simple words."
chat_prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI assistant."),
    ("human","Explain{topic}in simple words."),
])
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
# chain the prompt -> llm -> StrOutputParser
chain = chat_prompt | LLM | parser
# invoke the responce
result = chain.invoke({"topic":"Artificial Intelligence"})
print(result)
```

# Runnablesd
- It is the component that can recive an input and produce as output
- it can be used inside LCEL
- Runnabled parellel,runnableLambda, runnablepassthrough

```python
# Renabled Lamda example
from langchain_huggingface import HuggingFacePipeline
from transformers import pipeline
from langchain_core.prompts import ChatPromptTemplate
LLM = HuggingFacePipeline(pipeline=generator)

chat_prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI assistant."),
    ("human","Explain{topic}in simple words."),
])

from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()

from langchain_core.runnables import RunnableLambda
# now defiine the function which appeand strings "LLM responce is completed"
# to the output of LLM

def appeand_text(text):
  return text+"\nLLM responce is completed"

# now create runable lambda which calls above function
append_runnable = RunnableLambda(appeand_text)

# chain the prompt -> llm -> StrOutputParser
chain = chat_prompt | LLM | parser | append_runnable


# invoke the responce
result = chain.invoke({"topic":"Artificial Intelligence"})
print(result)
```
