```python
import pandas as pd
import numpy as np
```

## weather_data

```python

# Read dataset
df = pd.read_csv(r"weather_data - weather_data.csv")

print("First 5 Rows")
print(df.head())

print("\nShape:", df.shape)
print("\nColumns:", list(df.columns))

print("\nInformation")
df.info()

print("\nSummary Statistics")
print(df.describe(include='all'))

print("\nMissing Values")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Clean object columns
obj_cols = df.select_dtypes(include='object').columns
for col in obj_cols:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].replace(['', 'nan', 'NaN', 'None'], np.nan)
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Clean numeric columns
num_cols = df.select_dtypes(include=np.number).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())

print("\nMissing Values After Cleaning")
print(df.isnull().sum())

print("\nNumPy Statistics")
print("Mean:")
print(np.mean(df[num_cols], axis=0) if len(num_cols)>0 else "No numeric columns")
print("Median:")
print(np.median(df[num_cols], axis=0) if len(num_cols)>0 else "No numeric columns")
print("Standard Deviation:")
print(np.std(df[num_cols], axis=0) if len(num_cols)>0 else "No numeric columns")

out_file = "weather_data_cleaned.csv"
df.to_csv(out_file, index=False)
print("\nSaved cleaned dataset as:", out_file)
```
