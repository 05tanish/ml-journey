# Neural Language model on small corpus

1. convert the words (tokens) into tokens (integer) ids
2. create context target paird( input and output)
3. build embeddings layer for word representation
4. use NN to train and predicts

```python
sentences = [
    "i love matchine learning",
    "i love deep learning",
    "machine learning is useful",
    "machine learning is interesting"
    "deep learning is interesting",
    "deep learning is powerful",
    "artificial intelligence is useful",
    "artificial intelligence is powerful",
    "artificial intelligence is interesting",
]
for i in sentences:
    print(i)
```

```python
# build the ids to tokens
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer()
tokenizer.fit_on_texts(sentences)
#as we have token now, convert them to ids
word_index = tokenizer.word_index
vocab_size = len(word_index)+1
print(vocab_size)
print(word_index)
```

```python
# create input and output pairs
# context and target pair
# example "deep learning is powerful",
#(4, 1) target 2
#(1, 2) taget 9
```

```python
# creeate context-Target pairs
x=[]
y=[]
for sen in sentences:
    word  = sen.split()
    for i in range(2,len(word)):
        context = word[i-2:i]
        target = word[i]
        x.append([word_index[w] for w in context])
        y.append(word_index[target])

print(x)
print(y)

#next step for training on NN architecture covert list to np.array
```

```python
# covert to numpy array both x and y
import numpy as np
x = np.array(x)
y = np.array(y)
print(x.shape)
print(y.shape)
```

```python
from tensorflow.keras.layers import Embedding, Dense, Flatten
from tensorflow.keras.models import Sequential
model=Sequential([
     Embedding(input_dim=vocab_size,output_dim=16,input_length=2),
    Flatten(),
    Dense(units=32,activation="softmax")]
)
model.compile(loss="sparse_categorical_crossentropy",optimizer="adam",metrics=["accuracy"])
model.fit(x,y,epochs=100)
```
