```python
#stemming
from nltk.stem import PorterStemmer
stmmer = PorterStemmer()
word = ["playing", "plays", "study","studying"]
for wor in word:
  print(wor, stmmer.stem(wor))
```

```python
#stemization
nltk.download("wordnet")
from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
words = ["playing", "played", "studies", "cars"]
for wor in words:
  print(wor, lemm.lemmatize(wor))
```

```python
text = """
Dogs are running in the park.
The dogs played happily.
I am studying NLP and studying machine learning.
"""
#1 lower
text = text.lower()
#2tokeize
from nltk.tokenize import word_tokenize
wrtoken =  word_tokenize(text)
print (wrtoken)
#3stopwords removal
from nltk.corpus import stopwords
stopwords = set(stopwords.words("english"))
filtered_words = [
    word for word in wrtoken
    if word not in stopwords
]

print("\nAfter stop-word removal:")
print(filtered_words)


#4stemmization
from nltk.stem import PorterStemmer
stem = PorterStemmer()
for wor in filtered_words:
  print(wor, stem.stem(wor))
#lemitization

from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
for wor in filtered_words:
  print(lemm.lemmatize(wor))
```

```python
#bow
from sklearn.feature_extraction.text import CountVectorizer
documents = [
    "I love NLP",
    "I love machine learning",
    "NLP is interesting"
]
vectorize = CountVectorizer()
X = vectorize.fit_transform(documents)
print(vectorize.get_feature_names_out())
print(X.toarray())
```

```python
#tf-idf
from sklearn.feature_extraction.text import TfidfVectorizer
documents = [
    "I love Python",
    "I love NLP",
    "Python is powerful",
    "NLP is powerful"
]
transformers = TfidfVectorizer()
x = transformers.fit_transform(documents)
print(transformers.get_feature_names_out())
print(x.toarray())
```

```python
#cosine similarity
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
documents = [
    "I love Python",
    "I love NLP",
    "Python is powerful",
    "NLP is powerful"
]
vectorize = TfidfVectorizer()
x = vectorize.fit_transform(documents)

print(" checcking similarity ")
cosine = cosine_similarity(x)
print(cosine)

```
